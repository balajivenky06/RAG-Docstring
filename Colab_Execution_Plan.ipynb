```python
import subprocess
from google.colab import userdata

def commit_and_push(commit_msg):
    try:
        # Configure Git
        subprocess.run(["git", "config", "--global", "user.email", "balajivenky06@gmail.com"], check=True)
        subprocess.run(["git", "config", "--global", "user.name", "balajivenky06"], check=True)
        
        # Add and commit
        subprocess.run(["git", "add", "."], check=True)
        
        # Commit might fail if there are no changes, so we capture output and don't use check=True here
        result = subprocess.run(["git", "commit", "-m", commit_msg], capture_output=True, text=True)
        if "nothing to commit" in result.stdout:
            print(f"\u26a0\ufe0f No new changes to commit for: {commit_msg}")
            return
            
        # Get token from secrets (make sure you added it in your colab secrets tab!)
        try:
            github_token = userdata.get('github_token')
        except:
            print("\u26a0\ufe0f Secret 'github_token' not found in Colab. Using hardcoded token.")
            github_token = '<YOUR_GITHUB_PAT>'

        git_username = "balajivenky06"
        repo_url_with_pat = f"https://{git_username}:{github_token}@github.com/balajivenky06/RAG-Docstring.git"
        
        # Push to remote
        subprocess.run(["git", "push", repo_url_with_pat, "main"], check=True)
        print(f"\u2705 Successfully committed and pushed: {commit_msg}")
        
    except subprocess.CalledProcessError as e:
        print(f"\u274c Error during git operations: {e}")

# Usage Example:
# commit_and_push("phi plain model rerun results")
```
